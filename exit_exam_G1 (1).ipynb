##Data preprocessing

```python
for col in cat_cols:
    shop[col] = shop[col].fillna(shop[col].mode()[0])

print(shop)
```

```python
shop.isna().sum()
```

```python
shop["Timestamp"] = pd.to_datetime(shop["Timestamp"], dayfirst=True, errors="coerce")
shop["hour"] = shop["Timestamp"].dt.hour
shop["day"] = shop["Timestamp"].dt.day
shop["month"] = shop["Timestamp"].dt.month
shop = shop.drop(columns=["Timestamp"])
```

```python
shop= shop.drop([
    "Reason_Prefer_Online",
    "Reason_Prefer_InStore",
    "Online_Shopping_Satisfaction_Score",
    "Trust_In_Online_Reviews_Score",
    "Regret_After_Purchase_Frequency",
    "Review_Influence_Score"
], axis=1)
```

```python
shop['Price_Comparison_Frequency'] = shop['Price_Comparison_Frequency'].fillna(
    shop['Price_Comparison_Frequency'].median())

shop['Impulse_Purchase_Frequency'] = shop['Impulse_Purchase_Frequency'].fillna(
    shop['Impulse_Purchase_Frequency'].median())
```

```python
shop.isna().sum()
```

```python
for col in ["Social_Ads_Influence_Score","Discount_Importance_Score", "Likely_To_Buy_After_Ad_Score"]:
    shop[col] = shop[col].fillna(shop[col].median())

shop["Follows_Brands_On_Social"] = shop["Follows_Brands_On_Social"].fillna(
    shop["Follows_Brands_On_Social"].mode()[0])
```

```python
shop.isna().sum()
```

```python
num_col=shop.select_dtypes(include=np.number).columns
for col in num_col:
  sns.boxplot(y=shop[col])
  plt.title(f"Boxplot of {col}")
  plt.show()

```

```python
for col in num_col:
    q1 = shop[col].quantile(0.25)
    q3 = shop[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = shop[(shop[col] < lower) | (shop[col] > upper)]
    print(col, len(outliers))
```

```python
shop['Best_Shopping_Mode'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
y = shop["Best_Shopping_Mode"]
x = shop.drop(columns=["Best_Shopping_Mode"])
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
cols_to_clip=["Price_Comparison_Frequency","Impulse_Purchase_Frequency"]
q1 = x_train[cols_to_clip].quantile(0.25)
q3 = x_train[cols_to_clip].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

x_train[cols_to_clip] = x_train[cols_to_clip].clip(lower=lower, upper=upper,axis=1)
x_test[cols_to_clip] = x_test[cols_to_clip].clip(lower=lower, upper=upper,axis=1)
```

```python
y_train.value_counts()
y_test.value_counts()
```

```python
cat_col = x_train.select_dtypes(exclude=np.number).columns

for col in cat_col:
    print(col, x_train[col].nunique())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
from sklearn.preprocessing import OneHotEncoder
encode_cols = [
    "Gender",
    "Occupation_Status",
    "Country_Region",
    "Shop_Most_Where",
    "Preferred_Platform",
    "Primary_Device",
    "Top_Spending_Category",
    "Follows_Brands_On_Social",
    "Makes_Shopping_List",
    "Preferred_Payment_Method",
    "Uses_BNPL_Installments"
]
encoder=OneHotEncoder(handle_unknown="ignore",sparse_output=False)
x_train_encoded=encoder.fit_transform(x_train[encode_cols])
x_test_encoded=encoder.transform(x_test[encode_cols])
```

```python
x_train_encoded.shape
```

```python
num_cols=x_train.select_dtypes(include=np.number).columns
```

```python
x_train_num=x_train[num_cols]
x_test_num=x_test[num_cols]
```

```python
x_train = x_train.copy()
x_test = x_test.copy()

for col in x_train.columns:
    if pd.api.types.is_datetime64_any_dtype(x_train[col]):
        x_train[col] = x_train[col].dt.hour
        x_test[col] = x_test[col].dt.hour

for col in x_train.select_dtypes(include=["object", "string"]).columns:
    x_train[col] = x_train[col].astype("category").cat.codes
    x_test[col] = x_test[col].astype("category").cat.codes
```

```python
x_train = np.hstack([x_train.drop(columns=encode_cols).to_numpy(),
                          encoder.fit_transform(x_train[encode_cols])])

x_test = np.hstack([x_test.drop(columns=encode_cols).to_numpy(),
                         encoder.transform(x_test[encode_cols])])

from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
x_train_, y_train_= smote.fit_resample(x_train, y_train)
```

```python
x_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
x_train_scaled.shape
```

##Modeling


```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

model_lr=LogisticRegression()
model_lr.fit(x_train_scaled,y_train)
y_pred_lr=model_lr.predict(x_test_scaled)

print(accuracy_score(y_test,y_pred_lr))
print(confusion_matrix(y_test,y_pred_lr))
print(classification_report(y_test,y_pred_lr))
```

```python
from sklearn.tree import DecisionTreeClassifier

model_dt=DecisionTreeClassifier()
model_dt.fit(x_train_scaled,y_train)
y_pred_dt=model_dt.predict(x_test_scaled)

print(accuracy_score(y_test,y_pred_dt))
print(confusion_matrix(y_test,y_pred_dt))
print(classification_report(y_test,y_pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier

model_rf=RandomForestClassifier()
model_rf.fit(x_train_scaled,y_train)
y_pred_rf=model_rf.predict(x_test_scaled)

print(accuracy_score(y_test,y_pred_rf))
print(confusion_matrix(y_test,y_pred_rf))
print(classification_report(y_test,y_pred_rf))
```

```python
from xgboost import XGBClassifier
model_xgb=XGBClassifier()
model_xgb.fit(x_train_scaled,y_train)
y_pred_xgb=model_xgb.predict(x_test_scaled)

print(accuracy_score(y_test,y_pred_xgb))
print(confusion_matrix(y_test,y_pred_xgb))
print(classification_report(y_test,y_pred_xgb))
```

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn.fit(x_train_scaled, y_train)
y_pred_knn = model_knn.predict(x_test_scaled)

print(accuracy_score(y_test, y_pred_knn))
print(confusion_matrix(y_test, y_pred_knn))
print(classification_report(y_test, y_pred_knn))
```

```python
from sklearn.naive_bayes import GaussianNB

model_nb = GaussianNB()
model_nb.fit(x_train_scaled, y_train)
y_pred_nb = model_nb.predict(x_test_scaled)

print(accuracy_score(y_test, y_pred_nb))
print(confusion_matrix(y_test, y_pred_nb))
print(classification_report(y_test, y_pred_nb))
```

```python
from sklearn.model_selection import cross_val_score
models={
    "Logistic regression" :model_lr,
    "DecisionTree" :model_dt,
    "Random Forest" :model_rf,
    "xgboost" :model_xgb,
    "knn":model_knn,
    "Naive Bayes":model_nb
}
for name, model in models.items():
  scores=cross_val_score(
      model,
      x_train_scaled,
      y_train,
      cv=5,
      scoring="accuracy"
  )
  print(name)
  print(f"Scores: {scores}")
  print(f"Mean: {scores.mean()}")
  print()
```

```python
from sklearn.model_selection import GridSearchCV

param_grids={
    "n_estimators":[50,100,200],
    "max_depth":[None,5,10],
    "min_samples_split":[2,5]
}

grid_rf=GridSearchCV(
    model_rf,
    param_grids,
    cv=5,
    scoring="accuracy"
)
grid_rf.fit(x_train_scaled,y_train)
print(f"best parameters: {grid_rf.best_params_}")
print(f"best CV accuracy: {grid_rf.best_score_}")
```
